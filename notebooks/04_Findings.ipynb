## Business Context

Primeflix had 500K Customers that churned in a span of 90 days. These customers were mixed across the high, medium, and low consumption segments. To get these customers back, Primeflix decided to provide these customers with offers. There were 3 types of offers - Early Access, Free Trial, and Reduced subscription costs.

## Test Design
Out of 500K Customers, 10% were not marketed to as part of a control group. The remining 90% were split into 3 equal groups. During the first month, customers received the offer. Months 2-4 were the duration for which the offers were valid.

| **Offer** | **Share of Customers** |
|---|---|
| No Offer (Control) | 10% |
| Early Access | 30% |
| Free Trial | 30% |
| Discounted Rate | 30% |
|**Total**|**100%**|

## Key Findings


### 1. Did Marketing Work?
Yes, the marketing worked. The customers who received an offer resubscribed at atleast 47% more when compared to customers who did not receive an offer.

### 2. Which Offer Won by Segment?
Each segment had a different offer that won. 
    -   In the Low engagement segment, the *free trial* offer caused the best response (atleast 22% higher than other offers)
    -   In the Medium engagement segment, the *discount* offer caused the best response (atleast 17% higher than other offers)
    -   In the High engegemtn segment, the *early access* offer caused the best response (atleast 15% higher than the other offers)

### 3. Do Retained Customers Stay Engaged?
Retained customers do stay engaged, but their engagement levels do not depend on the offers they received. High segment customers continue to engage much more than Low engagement customers post resubscription


## Recommendation
Tailoring return offers to churned customers based on their past behavior is statistically the best strategy moving forward to encourage resubscription
