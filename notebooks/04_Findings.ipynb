## Business Context

Primeflix had 500K Customers that churned in a span of 90 days. These customers were mixed across the high, medium, and low consumption segments. Re-acquiring these customers is significantly cheaper than acquiring new ones, making re-engagement campaigns a high priority revenue recovery tool. To get these customers back, Primeflix decided to provide these customers with offers. There were 3 types of offers - Early Access, Free Trial, and Reduced subscription costs.

## Test Design
Out of 500K Customers, 10% were not marketed to as part of a control group. The remining 90% were split into 3 equal groups. During the first month, customers received the offer. Months 2-4 were the duration for which the offers were valid. After this, months 5-7 were observed to monitor the response of returned customers.

| **Offer** | **Share of Customers** |
|---|---|
| No Offer (Control) | 10% |
| Early Access | 30% |
| Free Trial | 30% |
| Discounted Rate | 30% |
|**Total**|**100%**|

## Key Findings


### 1. Did Marketing Work?
Yes, the marketing worked. Across all segments, the campaign drove between 91% and 192% more resubscriptions compared to customers who received no offer.

### 2. Which Offer Won by Segment?
Each segment had a different offer that won. 

| **Segment** | **Winning Offer** | **Lift over next best offer** |
|---|---|---|
| High | Early Access | 22% |
| Medium | Discounted Rate | 17% |
| Low | Free Trial | 31% |

### 3. Do Retained Customers Stay Engaged?
Retained customers do stay engaged, but their engagement levels do not depend on the offers they received. High segment customers continue to engage much more than Low engagement customers post resubscription. This means offer choice should be optimized for resubscription rate, not post-resubscription engagement — since engagement is driven by the customer's inherent viewing habits, not the incentive used to bring them back

## Recommendation
We recommend implementing a segment-based offer assignment system for all future re-engagement campaigns. Churned customers should be segmented by their pre-churn viewing behavior and assigned offers accordingly — Early Access for High engagement customers, Discounted Rate for Medium, and Free Trial for Low. This approach is statistically validated and directly actionable at the campaign setup stage.
